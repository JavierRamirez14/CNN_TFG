```python
from tensorflow.keras.datasets import mnist
import numpy as np

def to_one_hot(y, num_classes):
    return np.eye(num_classes)[y]

def crear_mini_batches(X, y, batch_size, shuffle=True):
    if shuffle:
        indices = np.arange(X.shape[0])
        np.random.shuffle(indices)
        X = X[indices]
        y = y[indices]

    mini_batches = []
    for i in range(0, X.shape[0], batch_size):
        X_mini_batch = X[i:i + batch_size]
        y_mini_batch = y[i:i + batch_size]
        mini_batches.append((X_mini_batch, y_mini_batch))

    return mini_batches

# Cargar datos MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalizar las imágenes a valores entre 0 y 1 y reestructurarlas a 2D
X_train = X_train.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0

# Aplanar las imágenes a un vector
X_train = X_train.reshape(X_train.shape[0], -1)
X_test = X_test.reshape(X_test.shape[0], -1)

# Convertir etiquetas a one-hot encoding
y_train = to_one_hot(y_train, 10)
y_test = to_one_hot(y_test, 10)

# Seleccionar 10 imágenes de cada dígito (0-9) para train (100 imágenes en total)
train_indices = []
for digit in range(10):
    indices = np.where(y_train[:, digit] == 1)[0]  # Encontrar índices de la clase actual
    train_indices.extend(indices[:10])  # Seleccionar las primeras 10 imágenes de la clase

X_train_sample = X_train[train_indices]
y_train_sample = y_train[train_indices]

# Seleccionar 10 imágenes de cada dígito (0-9) para test (100 imágenes en total)
test_indices = []
for digit in range(10):
    indices = np.where(y_test[:, digit] == 1)[0]  # Encontrar índices de la clase actual
    test_indices.extend(indices[:10])  # Seleccionar las primeras 10 imágenes de la clase

X_test_sample = X_test[test_indices]
y_test_sample = y_test[test_indices]

indices = np.arange(X_test_sample.shape[0])
np.random.shuffle(indices)
X_test_sample = X_test_sample[indices]
y_test_sample = y_test_sample[indices]

# Crear mini-batches con la muestra de datos
train_data = crear_mini_batches(X_train_sample, y_train_sample, batch_size=10)
test_data = crear_mini_batches(X_test_sample, y_test_sample, batch_size=10)

# Imprimir las formas de los datos de muestra
print("Forma de X_train_sample:", X_train_sample.shape)
print("Forma de y_train_sample:", y_train_sample.shape)
print("Forma de X_test_sample:", X_test_sample.shape)
print("Forma de y_test_sample:", y_test_sample.shape)
```
